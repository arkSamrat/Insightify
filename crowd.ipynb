```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_excel('crowd.xlsx')
```

```python
data.head()
```

```python
data.describe
```

```python
data.info(5)
```

```python
data.drop(columns='Special_Features',inplace = True)
```

```python
data.head()
```

```python
data['isWeekend'] = data['Day_of_Week'].isin(['Sunday', 'Saturday']).astype(int) 
```

```python
data.head(2)
```

```python
data['State'] = data['Place'].str.split(',').str[1].str.strip()
```

```python
data.head(2)
```

```python
# LinearRegressor
# LogisticRegressor
# DicisionTreeClassifire
# RandomForestClassifire
# XGBoost
# LightGBM
```

```python
data['Place'] = data['Place'].str.split(',').str[0].str.strip()
```

```python
data.head()
```

```python
data['Date'] = pd.to_datetime(data['Date'])
```

```python
data['Time']=pd.to_datetime(data['Time'],format='%H:%M:%S')
```

```python
data['hour'] = data['Time'].dt.hour
data['month'] = data['Date'].dt.month
data['day'] = data['Date'].dt.day
```

```python
data.head(2)
```

```python
data.drop(columns='Time',inplace = True, axis=1)
```

```python
data.head()
```

```python
data['Holiday'] = data['Holiday'].map({"Yes":1,"No":0})
```

```python
data['Weather'].unique()
```

```python
new_data = data.drop(columns='Date',axis = 1)
```

```python
new_data.head(2)
```

```python
all_categorical_columns = new_data.select_dtypes(include = ['object']).columns
```

```python
all_categorical_columns
```

```python
y=data['Crowd_Count (in Thousands)']
x= data.drop(columns = ['Crowd_Count (in Thousands)'])
```

```python
y = y/(max(y))
```

```python
y.head()
```

```python
# Column_Transformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer(
    transformers= [('cat',OneHotEncoder(),all_categorical_columns)]
)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.pipeline import Pipeline
```

```python
model = Pipeline(
    steps =[
        ('preprocess',preprocessor),
        ('regressor',LinearRegression())
    ]
)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.29,random_state=42 )
```

```python
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("MSE:", mse)
print("R2:", r2)
```
